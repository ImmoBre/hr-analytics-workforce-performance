## 02 - Data Quality Assessment

### Project Overview

Data quality assessment is a critical step before performing exploratory analysis. This notebook evaluates the completeness, consistency, and reliability of the dataset by identifying missing values, duplicate records, incorrect data types, and potential data quality issues.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", "{:,.2f}".format)

df = pd.read_csv("../data/HRDataset_v14.csv")
```

#### Dataset Shape

Display the number of rows and columns contained in the dataset.

```python
df.shape
```

#### Missing Values Analysis

Identify the number of missing values for every variable.

```python
missing_values = df.isnull().sum().sort_values(ascending=False)

missing_values
```

#### Missing Values Percentage

Calculate the percentage of missing values for each feature.

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

missing_percentage
```

#### Duplicate Records

Check whether duplicate observations exist in the dataset.

```python
df.duplicated().sum()
```

#### Data Types Review

Review the data type of every feature.

```python
df.dtypes
```

#### Unique Values

Determine the number of unique values for every column.

```python
df.nunique().sort_values()
```

#### Numerical Summary

Generate descriptive statistics for numerical variables.

```python
df.describe()
```

#### Categorical Summary

Generate descriptive statistics for categorical variables.

```python
df.describe(include="object")
```

#### Data Quality Conclusion

This assessment evaluates:

- Missing values
- Duplicate records
- Data types
- Unique values
- Overall dataset consistency

The findings from this notebook will guide the data preparation and exploratory analysis performed in the next stage of the project.

#### Final Assessment

The dataset demonstrates a high level of quality suitable for business analytics.

##### Key findings

- No duplicate observations were identified.
- Most variables contain complete information.
- The missing values in `DateofTermination` are expected because active employees have not been terminated.
- The small number of missing values in `ManagerID` represents only 2.57% of the dataset and does not materially impact the analysis.

Overall, the dataset is considered reliable for exploratory analysis, feature engineering, SQL business analysis, and dashboard development.
