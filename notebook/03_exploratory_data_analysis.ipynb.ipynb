# HR Analytics: Workforce Performance & Employee Insights

## 03 - Exploratory Data Analysis

### Project Overview

Exploratory Data Analysis (EDA) aims to discover meaningful patterns, trends, and relationships within the workforce dataset.

This notebook answers important business questions regarding employee distribution, salaries, performance, engagement, recruitment, and termination to support data-driven HR decision-making.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.2f}".format)

df = pd.read_csv("../data/HRDataset_v14.csv")
```

## Business Question 1

### How are employees distributed across departments?

Understanding workforce distribution across departments helps identify organizational structure and resource allocation.

```python
department_count = (
    df["Department"]
    .value_counts()
    .sort_values(ascending=False)
)

department_count
```

```python
plt.figure(figsize=(10,6))

department_count.plot(kind="bar")

plt.title("Employee Distribution by Department")

plt.xlabel("Department")

plt.ylabel("Number of Employees")

plt.xticks(rotation=45)

plt.show()
```

### Insight

This visualization highlights the departments employing the largest number of workers and provides an overview of workforce allocation across the organization.

### Business Insight

The Production department accounts for the majority of the workforce with 209 employees, representing a significant concentration of human resources within the organization.

The IT/IS department is the second largest with 50 employees, while Sales employs 31 individuals. Administrative and Executive functions represent only a small portion of the workforce.

This distribution suggests that the company is primarily operational and production-oriented, requiring a larger production workforce than administrative staff.

## Business Question 2

### What is the distribution of employee employment status?

Analyzing employment status helps understand the proportion of active and terminated employees.## Business Question 2

## Business Question 3

### Which departments have the highest average salary?

Understanding salary differences across departments helps identify compensation strategies and workforce investment.

```python
department_salary = (
    df.groupby("Department")["Salary"]
    .mean()
    .sort_values(ascending=False)
)

department_salary
```

```python
plt.figure(figsize=(10,6))

department_salary.plot(kind="bar")

plt.title("Average Salary by Department")

plt.xlabel("Department")

plt.ylabel("Average Salary")

plt.xticks(rotation=45)

plt.show()
```

### Business Insight

The Executive Office reports the highest average salary; however, this department consists of only one employee and should not be considered representative.

Among larger operational departments, IT/IS and Software Engineering have the highest average salaries, reflecting the specialized technical skills required for these roles.

Production, despite employing the majority of the workforce, has the lowest average salary, indicating a labor-intensive operational structure.

## Business Question 4

### Which job positions receive the highest average salary?

Analyzing salaries by position provides insight into compensation across organizational roles.

```python
position_salary = (
    df.groupby("Position")["Salary"]
    .mean()
    .sort_values(ascending=False)
)

position_salary
```

```python
plt.figure(figsize=(12,6))

position_salary.plot(kind="bar")

plt.title("Average Salary by Position")

plt.xlabel("Position")

plt.ylabel("Average Salary")

plt.xticks(rotation=90)

plt.show()
```

### Business Insight

Executive leadership positions such as President & CEO, CIO, and Director of Sales receive the highest compensation within the organization.

Technical leadership roles, including IT Director, Data Architect, and IT Managers, also rank among the highest-paid positions.

Conversely, operational roles such as Production Technician and Administrative Assistant receive comparatively lower salaries, reflecting differences in responsibility and specialization.

## Business Question 13

### What are the descriptive statistics of employee salaries?

Descriptive statistics summarize the central tendency and variability of employee compensation across the organization.

```python
salary_stats = pd.Series({
    "Mean": df["Salary"].mean(),
    "Median": df["Salary"].median(),
    "Mode": df["Salary"].mode()[0],
    "Minimum": df["Salary"].min(),
    "Maximum": df["Salary"].max(),
    "Standard Deviation": df["Salary"].std()
})

salary_stats
```

### Business Insight

The average employee salary is **$69,020.68**, while the median salary is **$62,810.00**.

Because the mean exceeds the median, the salary distribution appears positively skewed, indicating that a relatively small number of highly compensated employees increase the overall average.

The wide range between the minimum salary (**$45,046**) and maximum salary (**$250,000**) reflects the presence of executive leadership positions alongside operational roles.

## Business Question 14

### What are the descriptive statistics of employee absences?

Understanding absenteeism helps evaluate workforce reliability and operational efficiency.

```python
absence_stats = pd.Series({
    "Mean": df["Absences"].mean(),
    "Median": df["Absences"].median(),
    "Mode": df["Absences"].mode()[0],
    "Minimum": df["Absences"].min(),
    "Maximum": df["Absences"].max(),
    "Standard Deviation": df["Absences"].std()
})

absence_stats
```

The standard deviation of **5.85** indicates moderate variability, while the maximum of **20 absences** highlights that some employees experience substantially higher absenteeism than others.

Monitoring departments with elevated absence levels may help improve productivity and workforce planning.
