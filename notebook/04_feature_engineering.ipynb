# HR Analytics: Workforce Performance & Employee Insights

## 04 - Feature Engineering

### Project Overview

Feature engineering transforms existing variables into new business-oriented features that improve analytical capabilities and decision-making.

The engineered variables created in this notebook will support SQL analysis, Power BI dashboards, and executive business recommendations.

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/HRDataset_v14.csv")
```

## Feature 1

### Salary Category

Employees are categorized into salary groups to facilitate compensation analysis.

```python
df["SalaryCategory"] = pd.qcut(
    df["Salary"],
    q=4,
    labels=[
        "Low",
        "Medium",
        "High",
        "Executive"
    ]
)

df[
    [
        "Salary",
        "SalaryCategory"
    ]
].head()
```

## Feature 2

### Satisfaction Category

Employee satisfaction scores are converted into descriptive business categories.

```python
df["SatisfactionCategory"] = np.where(
    df["EmpSatisfaction"] >= 4,
    "High Satisfaction",
    np.where(
        df["EmpSatisfaction"] == 3,
        "Moderate Satisfaction",
        "Low Satisfaction"
    )
)

df[
    [
        "EmpSatisfaction",
        "SatisfactionCategory"
    ]
].head()
```

## Feature 3

### Engagement Category

Employee engagement scores are transformed into qualitative categories.

```python
df["EngagementCategory"] = np.where(
    df["EngagementSurvey"] >= 4,
    "Highly Engaged",
    np.where(
        df["EngagementSurvey"] >= 3,
        "Moderately Engaged",
        "Low Engagement"
    )
)

df[
    [
        "EngagementSurvey",
        "EngagementCategory"
    ]
].head()
```

## Feature 4

### Attendance Category

Employees are classified according to their absence levels.

```python
df["AttendanceCategory"] = np.where(
    df["Absences"] <= 5,
    "Excellent",
    np.where(
        df["Absences"] <= 10,
        "Good",
        np.where(
            df["Absences"] <= 15,
            "Average",
            "Poor"
        )
    )
)

df[
    [
        "Absences",
        "AttendanceCategory"
    ]
].head()
```

## Feature 5

### High Performer Flag

Employees are classified as high performers based on their performance evaluation.

```python
df["HighPerformer"] = np.where(
    df["PerformanceScore"] == "Exceeds",
    "Yes",
    "No"
)

df[
    [
        "PerformanceScore",
        "HighPerformer"
    ]
].head()
```

## Feature 6

### Attrition Flag

Employees are categorized based on whether they are currently active or have left the organization.

```python
df["AttritionFlag"] = np.where(
    df["EmploymentStatus"] == "Active",
    "Active",
    "Left"
)

df[
    [
        "EmploymentStatus",
        "AttritionFlag"
    ]
].head()
```

## Notebook Summary

Feature engineering created several business-oriented variables that simplify workforce segmentation and enhance analytical capabilities.

These engineered features will be leveraged in SQL business questions, Power BI dashboards, and executive strategic recommendations.
